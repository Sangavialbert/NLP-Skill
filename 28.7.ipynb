```python
# ============================================================
# MOVIE REVIEW SENTIMENT ANALYSIS USING NLP
# Dataset: IMDb Dataset of 50K Movie Reviews
# File Name: IMDB Dataset.csv
# ============================================================

# -----------------------------
# STEP 1: IMPORT LIBRARIES
# -----------------------------
import pandas as pd
import numpy as np
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# -----------------------------
# STEP 2: DOWNLOAD NLTK FILES
# -----------------------------
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

# -----------------------------
# STEP 3: LOAD DATASET
# -----------------------------
print("Loading Dataset...\n")

df = pd.read_csv("IMDB Dataset.csv")

print("First 5 Records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nSentiment Distribution:")
print(df['sentiment'].value_counts())

# -----------------------------
# STEP 4: TEXT PREPROCESSING
# -----------------------------
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()


def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove punctuation and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Tokenization
    words = word_tokenize(text)

    # Remove stopwords
    words = [word for word in words if word not in stop_words]

    # Lemmatization
    words = [lemmatizer.lemmatize(word) for word in words]

    # Join words
    return " ".join(words)


print("\nPreprocessing Reviews...")

df['clean_review'] = df['review'].apply(preprocess_text)

print("\nSample Cleaned Review:")
print(df['clean_review'][0])

# -----------------------------
# STEP 5: TF-IDF VECTORIZATION
# -----------------------------
print("\nApplying TF-IDF Vectorization...")

X = df['clean_review']
y = df['sentiment']

vectorizer = TfidfVectorizer(max_features=5000)

X = vectorizer.fit_transform(X)

print("Feature Matrix Shape:", X.shape)

# -----------------------------
# STEP 6: TRAIN-TEST SPLIT
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Samples:", X_train.shape[0])
print("Testing Samples :", X_test.shape[0])

# ============================================================
# MODEL 1: LOGISTIC REGRESSION
# ============================================================

print("\n" + "="*50)
print("LOGISTIC REGRESSION")
print("="*50)

lr_model = LogisticRegression(max_iter=1000)

lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)

print("\nAccuracy:",
      accuracy_score(y_test, lr_pred))

print("Precision:",
      precision_score(y_test, lr_pred,
                      pos_label='positive'))

print("Recall:",
      recall_score(y_test, lr_pred,
                   pos_label='positive'))

print("F1 Score:",
      f1_score(y_test, lr_pred,
              pos_label='positive'))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, lr_pred))

print("\nClassification Report:")
print(classification_report(y_test, lr_pred))

# ============================================================
# MODEL 2: NAIVE BAYES
# ============================================================

print("\n" + "="*50)
print("NAIVE BAYES")
print("="*50)

nb_model = MultinomialNB()

nb_model.fit(X_train, y_train)

nb_pred = nb_model.predict(X_test)

print("\nAccuracy:",
      accuracy_score(y_test, nb_pred))

print("Precision:",
      precision_score(y_test, nb_pred,
                      pos_label='positive'))

print("Recall:",
      recall_score(y_test, nb_pred,
                   pos_label='positive'))

print("F1 Score:",
      f1_score(y_test, nb_pred,
              pos_label='positive'))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, nb_pred))

print("\nClassification Report:")
print(classification_report(y_test, nb_pred))

# ============================================================
# MODEL 3: SUPPORT VECTOR MACHINE
# ============================================================

print("\n" + "="*50)
print("SUPPORT VECTOR MACHINE")
print("="*50)

svm_model = LinearSVC()

svm_model.fit(X_train, y_train)

svm_pred = svm_model.predict(X_test)

print("\nAccuracy:",
      accuracy_score(y_test, svm_pred))

print("Precision:",
      precision_score(y_test, svm_pred,
                      pos_label='positive'))

print("Recall:",
      recall_score(y_test, svm_pred,
                   pos_label='positive'))

print("F1 Score:",
      f1_score(y_test, svm_pred,
              pos_label='positive'))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, svm_pred))

print("\nClassification Report:")
print(classification_report(y_test, svm_pred))

# ============================================================
# STEP 7: PREDICT NEW REVIEWS
# ============================================================

print("\n" + "="*50)
print("PREDICTING NEW REVIEWS")
print("="*50)


def predict_sentiment(review):

    # Preprocess
    cleaned = preprocess_text(review)

    # Convert to TF-IDF
    vector = vectorizer.transform([cleaned])

    # Predict using Logistic Regression
    prediction = lr_model.predict(vector)

    return prediction[0]


# Example Reviews
review1 = """
This movie was amazing! The acting was brilliant,
and the storyline kept me engaged till the end.
"""

review2 = """
Terrible movie. Completely boring and a waste of
time. I would never recommend it to anyone.
"""

review3 = """
The movie was okay. Some scenes were good,
but overall it could have been much better.
"""

print("\nReview 1:")
print(review1)
print("Predicted Sentiment:",
      predict_sentiment(review1))

print("\nReview 2:")
print(review2)
print("Predicted Sentiment:",
      predict_sentiment(review2))

print("\nReview 3:")
print(review3)
print("Predicted Sentiment:",
      predict_sentiment(review3))

# ============================================================
# STEP 8: USER INPUT PREDICTION
# ============================================================

print("\n" + "="*50)
print("ENTER YOUR OWN REVIEW")
print("="*50)

while True:

    user_review = input(
        "\nEnter a Movie Review (or type 'exit'): "
    )

    if user_review.lower() == "exit":
        print("\nProgram Terminated.")
        break

    result = predict_sentiment(user_review)

    print("\nPredicted Sentiment:", result)

```
