```python
# ============================================================
# CUSTOMER REVIEW CLEANING SYSTEM
# IMDb 50K Review Dataset
# ============================================================

# Import libraries
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer


# ============================================================
# STEP 1: Download NLTK Resources
# ============================================================

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')


# ============================================================
# STEP 2: Load IMDb 50K Dataset using Pandas
# ============================================================

df = pd.read_csv("IMDB Dataset.csv")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Reviews:")
print(df.head())


# ============================================================
# STEP 3: Inspect Missing Values
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# STEP 4: Inspect Duplicate Reviews
# ============================================================

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("\nDataset Shape After Removing Duplicates:")
print(df.shape)


# ============================================================
# STEP 5: Initialize NLTK Tools
# ============================================================

stop_words = set(stopwords.words('english'))

lemmatizer = WordNetLemmatizer()


# ============================================================
# STEP 6: Create Review Cleaning Function
# ============================================================

def clean_review(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Remove numbers
    text = re.sub(r'\d+', ' ', text)

    # Remove punctuation and special characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenize the review
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # Apply lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    # Convert tokens back into a sentence
    cleaned_text = ' '.join(tokens)

    return cleaned_text


# ============================================================
# STEP 7: Apply Cleaning to All Reviews
# ============================================================

df['cleaned_review'] = df['review'].apply(clean_review)


# ============================================================
# STEP 8: Compare Original and Cleaned Reviews
# ============================================================

print("\n")
print("=" * 70)
print("ORIGINAL VS CLEANED REVIEWS")
print("=" * 70)

for i in range(5):

    print("\nOriginal:")
    print(df.iloc[i]['review'])

    print("\nCleaned:")
    print(df.iloc[i]['cleaned_review'])

    print("-" * 70)


# ============================================================
# STEP 9: Display Final Dataset
# ============================================================

print("\nFinal Dataset:")
print(df[['review', 'sentiment', 'cleaned_review']].head())


# ============================================================
# STEP 10: Export Preprocessed Dataset as CSV
# ============================================================

output_file = "IMDB_Preprocessed_Dataset.csv"

df[['review', 'sentiment', 'cleaned_review']].to_csv(
    output_file,
    index=False
)

print("\n")
print("=" * 70)
print("PREPROCESSING COMPLETED SUCCESSFULLY!")
print("=" * 70)

print("\nOutput File:", output_file)
```
